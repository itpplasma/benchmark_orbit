```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
confined_frac_data = np.loadtxt('../run_simple/precise_QA_fusion_distribution/npoints_256/confined_fraction.dat')

confined_frac = confined_frac_data[:, 1] + confined_frac_data[:, 2]
loss_frac_simple = 1 - confined_frac
time_simple = confined_frac_data[:, 0]
```

```python
import matplotlib.pyplot as plt

# Set global font size
plt.rcParams['font.size'] = 14 

firm3d_data = np.loadtxt('../run_firm3d/precise_QA_fusion_distribution/loss_frac.txt')
time_firm3d = firm3d_data[:, 0]
loss_frac_firm3d = firm3d_data[:, 1]

plt.semilogy(time_simple, loss_frac_simple, label='Simple')
plt.semilogy(time_firm3d, loss_frac_firm3d, label='Firm3D')
plt.ylim([1e-3,2e-2])
plt.legend()
plt.xlabel('Time [s]')
plt.ylabel('Loss Fraction')
plt.savefig('loss_fraction_precise_QA.png', dpi=300)
plt.show()
```
